```python
import os
from google.colab import userdata
from crewai import Agent, Task, Crew, LLM
from crewai.tools import tool # This is our "Adapter"
from langchain_community.tools import DuckDuckGoSearchRun
```

```python
# 1. Setup API Key
!pip install litellm
os.environ["GROQ_API_KEY"] = userdata.get('GROQ_API_KEY')
my_llm = LLM(model="groq/llama-3.3-70b-versatile")
#os.environ["GOOGLE_API_KEY"] = userdata.get('GOOGLE_API_KEY')
#my_llm = LLM(model="gemini/gemini-2.0-flash")
#my_llm = LLM(model="gemini/gemini-1.5-flash")
```

```python
# 2. Create the Wrapper (The Adapter)
# This @tool decorator tells CrewAI: "Treat this function as a valid tool"
@tool("internet_search")
def internet_search(query: str):
    """Searches the internet for a given topic and returns the results."""
    search = DuckDuckGoSearchRun()
    return search.run(query)
```

```python
# 3. Define the Agent
researcher = Agent(
  role='Tech Researcher',
  goal='Find the top AI news',
  backstory='Expert analyst.',
  tools=[internet_search], # We use our new adapter here
  llm=my_llm,
  max_iter=3,
  verbose=True
)
```

```python
# 4. Define the Task
task = Task(
  description='Summarize 3 major AI news stories from this week.',
  expected_output='A bulleted list of 3 news items.',
  agent=researcher
)
```

# Lets Gooooooo

```python
# 5. Kickoff
crew = Crew(agents=[researcher], tasks=[task])
print(crew.kickoff())
```
